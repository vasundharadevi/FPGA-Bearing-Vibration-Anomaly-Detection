```python
import numpy as np
import matplotlib.pyplot as plt

# Select vibration channel 1 (column 0)
signal = df.iloc[:, 0].values

# Number of samples in each window
WINDOW_SIZE = 256

# Calculate number of complete windows
num_windows = len(signal) // WINDOW_SIZE

# Store MAV values
mav_values = []

for i in range(num_windows):

    # Extract 256 samples
    window = signal[i * WINDOW_SIZE : (i + 1) * WINDOW_SIZE]

    # Calculate Mean Absolute Value
    mav = np.mean(np.abs(window))

    mav_values.append(mav)

mav_values = np.array(mav_values)

print("Total samples:", len(signal))
print("Window size:", WINDOW_SIZE)
print("Number of windows:", len(mav_values))
print("First 10 MAV values:", mav_values[:10])
```

```python
plt.figure(figsize=(12, 4))

plt.plot(mav_values, marker='o', markersize=3)

plt.title("Mean Absolute Value (MAV) - Vibration Signal")
plt.xlabel("Window Number")
plt.ylabel("MAV")
plt.grid(True)

plt.show()
```

```python
import os

folder_path = os.path.join(
    dataset_path,
    "1st_test",
    "1st_test"
)

data_files = sorted([
    os.path.join(folder_path, f)
    for f in os.listdir(folder_path)
    if os.path.isfile(os.path.join(folder_path, f))
])

# Select the first 20 snapshots
selected_files = data_files[:20]

print("Total available files:", len(data_files))
print("Files selected:", len(selected_files))
```

```python
import numpy as np
import pandas as pd

WINDOW_SIZE = 256

all_mav_values = []

for file in selected_files:

    # Load one measurement snapshot
    data = pd.read_csv(
        file,
        sep=r"\s+",
        header=None
    )

    # Extract channel 1
    signal = data.iloc[:, 0].values

    # Divide into complete windows
    num_windows = len(signal) // WINDOW_SIZE

    # Calculate MAV for each window
    for i in range(num_windows):

        window = signal[
            i * WINDOW_SIZE : (i + 1) * WINDOW_SIZE
        ]

        mav = np.mean(np.abs(window))

        all_mav_values.append(mav)

all_mav_values = np.array(all_mav_values)

print("Total snapshots:", len(selected_files))
print("Total MAV values:", len(all_mav_values))
print("Mean MAV:", np.mean(all_mav_values))
print("Standard deviation:", np.std(all_mav_values))
print("Minimum MAV:", np.min(all_mav_values))
print("Maximum MAV:", np.max(all_mav_values))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

plt.plot(all_mav_values)

plt.title("MAV Across 20 Bearing Measurement Snapshots")
plt.xlabel("Window Number")
plt.ylabel("MAV")
plt.grid(True)

plt.show()
```

```python
baseline_mean = np.mean(all_mav_values)
baseline_std = np.std(all_mav_values)

threshold = baseline_mean + (3 * baseline_std)

print("Baseline Mean:", baseline_mean)
print("Baseline Standard Deviation:", baseline_std)
print("Provisional Threshold:", threshold)
```

```python
abnormal_count = np.sum(all_mav_values > threshold)
normal_count = np.sum(all_mav_values <= threshold)

print("Normal windows:", normal_count)
print("Abnormal windows:", abnormal_count)

print(
    "Percentage above threshold:",
    abnormal_count / len(all_mav_values) * 100,
    "%"
)
```

```python
# Identify MAV values exceeding the threshold
abnormal_indices = np.where(all_mav_values > threshold)[0]

print("Windows exceeding threshold:")
print("-----------------------------")

for index in abnormal_indices:
    snapshot_number = index // 80 + 1
    window_number = index % 80 + 1

    print(
        f"Snapshot: {snapshot_number}, "
        f"Window: {window_number}, "
        f"MAV: {all_mav_values[index]:.6f}"
    )
```

```python
plt.figure(figsize=(14, 5))

plt.plot(
    all_mav_values,
    label="MAV values",
    linewidth=1
)

plt.axhline(
    y=threshold,
    linestyle="--",
    label="Provisional threshold"
)

plt.scatter(
    abnormal_indices,
    all_mav_values[abnormal_indices],
    color="red",
    label="Threshold exceedances",
    zorder=3
)

plt.title("Bearing Vibration Anomaly Detection")
plt.xlabel("Window Number")
plt.ylabel("MAV")
plt.legend()
plt.grid(True)

plt.show()
```

```python
snapshot_mean_mav = []
snapshot_max_mav = []

for file in selected_files:

    data = pd.read_csv(
        file,
        sep=r"\s+",
        header=None
    )

    signal = data.iloc[:, 0].values

    num_windows = len(signal) // WINDOW_SIZE

    snapshot_mav = []

    for i in range(num_windows):

        window = signal[
            i * WINDOW_SIZE : (i + 1) * WINDOW_SIZE
        ]

        mav = np.mean(np.abs(window))
        snapshot_mav.append(mav)

    snapshot_mean_mav.append(np.mean(snapshot_mav))
    snapshot_max_mav.append(np.max(snapshot_mav))

print("Snapshot analysis completed!")
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    range(1, 21),
    snapshot_mean_mav,
    marker="o",
    label="Average MAV"
)

plt.axhline(
    baseline_mean,
    linestyle="--",
    label="Overall baseline mean"
)

plt.title("Average Vibration Level Across 20 Snapshots")
plt.xlabel("Snapshot Number")
plt.ylabel("Average MAV")
plt.xticks(range(1, 21))
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    range(1, 21),
    snapshot_max_mav,
    marker="o",
    label="Maximum MAV"
)

plt.axhline(
    threshold,
    linestyle="--",
    label="Provisional threshold"
)

plt.title("Maximum Vibration Level Across 20 Snapshots")
plt.xlabel("Snapshot Number")
plt.ylabel("Maximum MAV")
plt.xticks(range(1, 21))
plt.legend()
plt.grid(True)

plt.show()
```
